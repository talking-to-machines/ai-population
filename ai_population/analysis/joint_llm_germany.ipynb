```python
import pandas as pd
stock_pick_data = pd.read_csv("../data/market-signals-x/test/x_finfluencer_daily_stock_pick_test_full.csv")
print(stock_pick_data.shape)
stock_pick_data["account_id"].info()

```

```python
missing = stock_pick_data.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print(missing)
```

```python
missing_df = pd.DataFrame({
    "missing_count": missing,
    "missing_pct": (missing / len(stock_pick_data)) * 100
})
print(missing_df)
```
