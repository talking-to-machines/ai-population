```python
metadata = pd.read_csv("../data/digital-twin-chile-x/raw_data/final_meta_user_df.csv")
print(metadata.shape)
metadata.head()
```

```python
tweets = pd.read_csv("../data/digital-twin-chile-x/raw_data/final_tweets_user_df.csv")
print(tweets.shape)
tweets.head()
```

```python
account_list = [account for account in metadata["userName"].unique() if account in tweets["author_username"].unique()]
account_list = account_list[:5]
len(account_list)
```

```python
filtered_metadata = metadata[metadata["userName"].isin(account_list)].reset_index(drop=True)
filtered_metadata.rename(columns={
    "profilePicture":"profile_picture",
    "userName":"account_id",
    # "createdAt":"created_at",
    "isVerified":"is_verified",
    "isBlueVerified":"is_blue_verified",
    "statusesCount":"statuses_count",
    "favouritesCount":"favourites_count",
    "mediaCount":"media_count",
}, 
inplace=True)
filtered_metadata.to_csv("../data/digital-twin-chile-x/profile_metadata_test.csv", index=False)
filtered_metadata.head()
```

```python
filtered_tweets = tweets[tweets["author_username"].isin(account_list)].reset_index(drop=True)
filtered_tweets.rename(columns={
    "tweet_created_at":"createdAt",
    "tweet_text":"text",
    "like_count":"likeCount",
    "view_count":"viewCount",
    "retweet_count":"retweetCount",
    "reply_count":"replyCount",
    "quote_count":"quoteCount",
    "bookmark_count":"bookmarkCount",
    "tweet_lang":"lang",
    "author_username":"account_id",
}, 
inplace=True)
filtered_tweets.to_csv("../data/digital-twin-chile-x/profile_tweets_test.csv", index=False)
filtered_tweets.head()
```
