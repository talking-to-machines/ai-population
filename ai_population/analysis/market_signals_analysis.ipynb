```python
# Load the finfluencer pool and historical profile search data
finfluencer_pool = pd.read_csv(f"../data/market-signals-x/{FINFLUENCER_POOL_FILE_X}")
finfluencer_pool = finfluencer_pool[
        finfluencer_pool["account_id"].isin(
            DAILY_STOCK_PICK_PROFILES_X + ORIGINAL_PROFILES_X
        )
    ]
historical_posts = pd.read_csv(f"../data/market-signals-x/{FINFLUENCER_HISTORICAL_PROFILE_SEARCH_FILE_X}")

print(f"Finfluencer pool size: {len(finfluencer_pool)}")
print(f"Total historical posts: {len(historical_posts)}")
print(f"Unique accounts in historical posts: {historical_posts['account_id'].nunique()}")
```

```python
# Filter historical posts to only include accounts from the finfluencer pool
pool_account_ids = set(finfluencer_pool['account_id'].tolist())
filtered_posts = historical_posts[historical_posts['account_id'].isin(pool_account_ids)]

print(f"Posts from finfluencer pool accounts: {len(filtered_posts)}")
print(f"Accounts from pool found in historical data: {filtered_posts['account_id'].nunique()}")
```

```python
# Count posts per account
posts_per_account = filtered_posts.groupby('account_id').size().reset_index(name='post_count')

# Merge with pool data to include accounts with 0 posts
posts_distribution = finfluencer_pool[['account_id']].merge(
    posts_per_account, 
    on='account_id', 
    how='left'
).fillna(0)
posts_distribution['post_count'] = posts_distribution['post_count'].astype(int)

# Summary statistics
print("=== Post Distribution Summary ===")
print(f"Total accounts in pool: {len(posts_distribution)}")
print(f"Accounts with posts: {(posts_distribution['post_count'] > 0).sum()}")
print(f"Accounts without posts: {(posts_distribution['post_count'] == 0).sum()}")
print(f"\nPost count statistics:")
print(posts_distribution['post_count'].describe())
```

```python
import matplotlib.pyplot as plt

# Histogram of post distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(posts_distribution['post_count'], bins=30, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Number of Posts')
axes[0].set_ylabel('Number of Accounts')
axes[0].set_title('Distribution of Posts per Account')
axes[0].axvline(posts_distribution['post_count'].median(), color='red', linestyle='--', label=f"Median: {posts_distribution['post_count'].median():.0f}")
axes[0].axvline(posts_distribution['post_count'].mean(), color='green', linestyle='--', label=f"Mean: {posts_distribution['post_count'].mean():.1f}")
axes[0].legend()

# Box plot
axes[1].boxplot(posts_distribution['post_count'], vert=True)
axes[1].set_ylabel('Number of Posts')
axes[1].set_title('Box Plot of Posts per Account')

plt.tight_layout()
plt.show()
```

```python
# Percentile distribution
import numpy as np

percentiles = [10, 25, 50, 75, 90, 95, 99]
print("=== Percentile Distribution ===")
for p in percentiles:
    value = np.percentile(posts_distribution['post_count'], p)
    print(f"{p}th percentile: {value:.0f} posts")

# Binned distribution (how many accounts fall into each range)
bins = [0, 10, 50, 100, 200, 500, 1000, float('inf')]
labels = ['0-10', '11-50', '51-100', '101-200', '201-500', '501-1000', '1000+']
posts_distribution['bin'] = pd.cut(posts_distribution['post_count'], bins=bins, labels=labels, right=True)

print("\n=== Binned Distribution ===")
bin_counts = posts_distribution['bin'].value_counts().sort_index()
for bin_label, count in bin_counts.items():
    pct = count / len(posts_distribution) * 100
    print(f"{bin_label}: {count} accounts ({pct:.1f}%)")
```
